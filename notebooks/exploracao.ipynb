## Verificação de linhas duplicadas (test):

```python
linhas_duplicadas =df_test.duplicated().sum()

if linhas_duplicadas > 0:
    print(f"Existem {linhas_duplicadas} linhas duplicadas no DataFrame.")
else:
    print("Não existem linhas duplicadas no DataFrame Test.")    
```

## Verificação de caracteres especiais em colunas que deveriam ser numéricas (test):

```python
colunas_deveriam_ser_numericas = ['Age', 'Annual_Income', 'Num_of_Loan', 'Num_of_Delayed_Payment', 'Changed_Credit_Limit', 'Outstanding_Debt', 'Amount_invested_monthly', 'Monthly_Balance']

for coluna in colunas_deveriam_ser_numericas:
    caracteres_especiais = df_train[df_train[coluna].astype(str).str.contains(r'[^0-9]', na=False)][coluna].unique()

    print(f"Amostra de caracteres especiais encontrados na coluna {coluna}:\n")
    print(caracteres_especiais[:5], "\n")
```

```python
# Descobrir o nome da coluna com problema de tipo misto
nome_coluna_problema = df_train.columns[26] # Pega a última coluna
print(f"A coluna com tipos mistos é: {nome_coluna_problema}")

# Ver que tipos de dados estão misturados nela
print(df_train[nome_coluna_problema].apply(type).value_counts())
```

```python
# 1. Conta quantos IDs repetidos existem no total
total_duplicados = df_test['Customer_ID'].duplicated().sum()

print("=" * 50)
print(f"🔍 ANÁLISE DE IDs REPETIDOS")
print("=" * 50)
print(f"• Total de linhas no dataset: {len(df_test)}")
print(f"• Quantidade de IDs repetidos: {total_duplicados}")

# 2. Se existirem duplicados, vamos investigar o motivo
if total_duplicados > 0:
    print("\n🚨 ATENÇÃO: Existem IDs duplicados!")
    print("\nExemplo das primeiras linhas com IDs repetidos para análise:")
    
    # Filtra o dataframe para mostrar as linhas que têm IDs duplicados
    # keep=False faz o Pandas mostrar todas as cópias daquele ID para você comparar
    IDs_repetidos = df_test[df_test['Customer_ID'].duplicated(keep=False)].sort_values(by='Customer_ID')
    
    # Mostra colunas estratégicas para você entender o motivo (ex: se muda o mês)
    colunas_para_olhar = ['ID', 'Customer_ID', 'Month', 'Age'] 
    # (Se alguma dessas colunas não existir exatamente com esse nome, ajuste aqui)
    
    display(IDs_repetidos[colunas_para_olhar].head(6))
else:
    print("\n✔ Excelente! Todos os IDs são únicos no dataset.")
print("=" * 50)
```
